```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.models import Sequential, Model
from tensorflow.keras.layers import Dense, Input, Dropout
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore', module='tensorflow')
```

```python
input_layer = Input(shape=(20,))           #Generate the input layer
print(input_layer)
layer1 = Dense(64, activation='relu')(input_layer)
print(layer1)
dropout_layer = Dropout(0.7)(layer1) #Adding a dropout layer
print(dropout_layer)
layer2 = Dense(64, activation='relu')(dropout_layer)
print(layer2)
output_layer = Dense(1, activation='sigmoid')(layer2)
print(output_layer)
```

```python
model = Model(inputs=input_layer, outputs=output_layer)
model.summary() #Provides a summary of the model created by showing the layers and the number of parameters
```

```python
#Compile the model created
model.compile(optimizer='adam', loss='mse', metrics=['accuracy'])
```

```python
#Train the model

X_train = np.random.rand(1000, 20)
Y_train = np.random.randint(2, size=(1000, 1))

model.fit(X_train, Y_train, epochs=100, batch_size=32)
```

```python
#Evaluate the model

X_test = np.random.rand(100, 20)
Y_test = np.random.randint(2, size=(100, 1))

loss, accuracy = model.evaluate(X_test, Y_test)
print(f"Test loss: {loss}")
print(f"Test accuracy: {accuracy}")
```

```python
#BUILDING CUSTOM LAYER

class CustomDense(layers.Layer):
  def __init__(self, units=32):
    super(CustomDense, self).__init__()
    self.units = units

  def build(self, input_shape):
    self.w = self.add_weight(shape=(input_shape[-1], self.units),
                             initializer='random_normal',
                             trainable=True)
    self.b = self.add_weight(shape=(self.units,),
                             initializer='zeros',
                             trainable=True)

  def call(self, inputs):
    output = tf.matmul(inputs, self.w) + self.b

    if self.activation =='relu':
      output = tf.nn.relu(output)
    return output

```

```python
from tensorflow.keras.layers import Softmax

model = Sequential([
    CustomDense(64),
    Dropout(0.7),
    CustomDense(64),
    Softmax()])

model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

print("Model summary before training:")
model.summary()
```
